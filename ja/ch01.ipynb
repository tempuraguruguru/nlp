## 00. パタトクカシーー
2つの文字列「パトカー」と「タクシー」の文字を先頭から交互に連結し、文字列「パタトクカシーー」を得よ。

```python
s1 = "パトカー"
s2 = "タクシー"
s = ""

for c1, c2 in zip(s1, s2):
  s += c1 + c1

s
```

## 04. 元素記号
"Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."という文を単語に分解し、1, 5, 6, 7, 8, 9, 15, 16, 19番目の単語は先頭の1文字、それ以外の単語は先頭の2文字を取り出し、取り出した文字列から単語の位置（先頭から何番目の単語か）への連想配列（辞書型もしくはマップ型）を作成せよ。

```python
import re
```

```python
S = "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."
S = re.sub('\.|\,', '', S)
list_S = re.split(' ', S)

dic_S = {}
for i, ele in enumerate(list_S):
  if ele[0] not in dic_S:
    dic_S[ele[0]] = i+1
  else:
    dic_S[ele[0:2]] = i+1

dic_S
```

## 05. n-gram
与えられたシーケンス（文字列やリストなど）からn-gramを作る関数を作成せよ。この関数を用い、"I am an NLPer"という文から文字tri-gram、単語bi-gramを得よ。

```python
import re
```

```python
def GenerateCharTrigram(S: str) -> list:
  res = []
  for i in range(len(S) - 2):
    res.append(S[i:i+3])
  return res

def GenerateWordBigram(S: str) -> list:
  res = []
  list_S = re.split(' ', S)
  for i in range(len(list_S) - 1):
    res.append(' '.join(list_S[i:i+2]))
  return res
```

```python
S = "I am an NLPer"
print(GenerateCharTrigram(S))
print(GenerateWordBigram(S))
```

## 06. 集合
"paraparaparadise"と"paragraph"に含まれる文字bi-gramの集合を、それぞれ, $X$ と $Y$ として求め、$X$ と $Y$ の和集合（$X \cup Y$）、積集合（$X \cap Y$）、差集合（$X \setminus Y$）を求めよ。さらに、'se'というbi-gramがXおよびYに含まれるかどうかを調べよ。

```python
import re
```

```python
def GenerateCharBigram(S: str) -> list:
  res = []
  for i in range(len(S)-1):
    res.append(S[i:i+2])
  return res
```

```python
S1 = set(GenerateCharBigram("paraparaparadise"))
S2 = set(GenerateCharBigram("paragraph"))
print(S1)
print(S2)
```

```python
print(S1 | S2) # 和集合
print(S1 & S2) # 積集合
print(S1 - S2) # 差集合
print('se' in (S1 | S2))
```

## 08. 暗号文
与えられた文字列の各文字を、以下の仕様で変換する関数cipherを実装せよ。

* 英小文字ならば (219 - 文字コード) のASCIIコードに対応する文字に置換
* その他の文字はそのまま出力

この関数を用い、英語のメッセージを暗号化・復号化せよ。

```python
def cipher(S: str) -> str:
  res = ""
  for c in S:
    if c.islower():
      res += chr(219 - ord(c))
    else:
      res += c
  return res
```

```python
def decryption(S: str) -> str:
  res = ""
  for c in S:
    if c.islower():
      res += chr(219 - ord(c))
    else:
      res += c
  return res
```

```python
S = "I can fly."
S_cipher = cipher(S)
S_decryption = decryption(S_cipher)
print(S_cipher)
print(S_decryption)
```

## 09. Typoglycemia
スペースで区切られた単語列に対して、各単語の先頭と末尾の文字は残し、それ以外の文字の順序をランダムに並び替えるプログラムを作成せよ。ただし、長さが４以下の単語は並び替えないこととする。適当な英語の文（例えば"I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."）を与え、その実行結果を確認せよ。

```python
import random
```

```python
def MyShuffle(S: str) -> str:
  res = ""
  dic_S = {}
  Shuffle_S = []
  list_S = S.split(' ')
  for i, ele in enumerate(list_S):
    if i == 0 or i == len(list_S) - 1 or len(ele) <= 4:
      dic_S[i] = ele
    else:
      dic_S[i] = ""
      Shuffle_S.append(ele)
  random.shuffle(Shuffle_S)
  for key, value in dic_S.items():
    if value == "":
      dic_S[key] = Shuffle_S.pop()
  for key, value in dic_S.items():
    res += value + " "
  return res.rstrip(" ")
```

```python
S = "I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."
print(MyShuffle(S))
```
