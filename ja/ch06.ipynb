## 54. アナロジーデータでの実験

[単語アナロジーの評価データ](http://download.tensorflow.org/data/questions-words.txt)をダウンロードし、国と首都に関する事例（`: capital-common-countries`セクション）に対して、vec(2列目の単語) - vec(1列目の単語) + vec(3列目の単語)を計算し、そのベクトルと類似度が最も高い単語と、その類似度を求めよ。求めた単語と類似度は、各事例と一緒に記録せよ。

```python
import pandas as pd
import re
```

```python
data = {'Word1': [], 'Word2': [], 'Word3': [], 'Word4(Correct)': [], 'Category': []}
with open('/content/drive/MyDrive/nlp100/lesson06/questions-words.txt', 'r') as file:
  category = ''
  for line in file:
    line = line.rstrip('\n')
    if ':' in line:
      category = re.sub('\s*:\s*(.+)', r'\1', line)
      continue
    else:
      words = re.split('\s+', line)
      data['Word1'].append(words[0])
      data['Word2'].append(words[1])
      data['Word3'].append(words[2])
      data['Word4(Correct)'].append(words[3])
      data['Category'].append(category)
df = pd.DataFrame(data)
```

```python
df_capital_common_countries = df[df['Category'] == 'capital-common-countries']

similary_word = []
similary_value = []
for index, row in df_capital_common_countries.iterrows():
  if index % 999 == 0:
    print(f'== {index}行目 ==')
  most_similar_info = model.most_similar(model[row[1]] - model[row[0]] + model[row[2]], topn = 1)[0]
  # most_similar_info = model.most_similar(positive = [row[0], row[2]], negative = [row[1]], topn = 1)[0] # 推奨
  similary_word.append(most_similar_info[0])
  similary_value.append(most_similar_info[1])

similary_word = pd.Series(similary_word)
similary_value = pd.Series(similary_value)
df_capital_common_countries['Similar_Word'] = similary_word
df_capital_common_countries['Similar_Value'] = similary_value
df_capital_common_countries
```

```python
df_capital_common_countries.to_csv('/content/drive/MyDrive/nlp100/lesson06/df_evaluated.csv', index = False)
```

## 55. アナロジータスクでの正解率

54の実行結果を用い、意味的アナロジー（semantic analogy）と文法的アナロジー（syntactic analogy）の正解率を測定せよ。

```python
df_semantic = df[~df['Category'].str.contains('gram')] # 意味的アナロジー

similary_word = []
similary_value = []
for index, row in df_semantic.iterrows():
  if index % 999 == 0:
    print(f'== {index}行目 ==')
  # most_similar_info = model.most_similar(model[row[1]] - model[row[0]] + model[row[2]], topn = 1)[0]
  most_similar_info = model.most_similar(positive = [row[1], row[2]], negative = [row[0]], topn = 1)[0] # 推奨
  similary_word.append(most_similar_info[0])
  similary_value.append(most_similar_info[1])

similary_word = pd.Series(similary_word)
similary_value = pd.Series(similary_value)
df_semantic['Similar_Word'] = similary_word
df_semantic['Similar_Value'] = similary_value
df_semantic
```

```python
df_syntactic = df[df['Category'].str.contains('gram')] # 文法的アナロジー

similary_word = []
similary_value = []
for index, row in df_syntactic.iterrows():
  if index % 999 == 0:
    print(f'== {index}行目 ==')
  # most_similar_info = model.most_similar(model[row[1]] - model[row[0]] + model[row[2]], topn = 1)[0]
  most_similar_info = model.most_similar(positive = [row[1], row[2]], negative = [row[0]], topn = 1)[0] # 推奨
  similary_word.append(most_similar_info[0])
  similary_value.append(most_similar_info[1])

similary_word = pd.Series(similary_word)
similary_value = pd.Series(similary_value)
df_syntactic['Similar_Word'] = similary_word
df_syntactic['Similar_Value'] = similary_value
df_syntactic
```

```python
df_semantic.to_csv('/content/drive/MyDrive/nlp100/lesson06/df_semantic.csv', index = False)
df_syntactic.to_csv('/content/drive/MyDrive/nlp100/lesson06/df_syntactic.csv', index = False)
```

```python
corrects = 0
for index, row in df_semantic.iterrows():
  if row['Similar_Word'] == row['Word4(Correct)']:
    corrects += 1
print(f'意味的アナロジーの正解率: {corrects / len(df_semantic)}')
```

## 56. WordSimilarity-353での評価

[The WordSimilarity-353 Test Collection](http://www.gabrilovich.com/resources/data/wordsim353/wordsim353.html)の評価データをダウンロードし、単語ベクトルにより計算される類似度のランキングと、人間の類似度判定のランキングの間のスピアマン相関係数を計算せよ。

```python
from scipy.stats import spearmanr
```

```python
df = pd.read_csv('/content/drive/MyDrive/nlp100/lesson06/wordsim353/combined.csv')
similary_value = []
for index, row in df.iterrows():
  similary_value.append(model.similarity(row['Word 1'], row['Word 2']))
df['System'] = pd.Series(similary_value)
print(df[['Human (mean)', 'System']].corr(method = 'spearman')) # methodのdefaultはpearson,
```

```python
# Spearmanの順位相関係数とp値を計算
corr, p_value = spearmanr(df['Human (mean)'], df['System'])

print(f"スピアマンの順位相関係数: {corr}")
print(f"p値: {p_value}")

# 統計的有意性の判断
if p_value < 0.05:
    print("相関は統計的に有意です（p < 0.05）。")
else:
    print("相関は統計的に有意ではありません（p >= 0.05）。")
```
