## 10. 行数のカウント

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  count = 0
  for line in file:
    count += 1
print(count)
```

## 11. 先頭からN行を出力
ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    print(lines[i])
```

## 12. 末尾のN行を出力
ファイルの末尾N行だけを表示せよ。例えば、N=10として末尾10行を表示せよ。確認にはtailコマンドを用いよ。

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    print(lines[-i])
```

## 13. タブをスペースに置換
ファイルの先頭10行に対して、タブ1文字につきスペース1文字に置換して出力せよ。確認にはsedコマンド、trコマンド、もしくはexpandコマンドなどを用いよ。

```python
import re
```

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    line = lines[i]
    line = re.sub('\t', ' ', line)
    print(line)
```

## 14. 1列目を出力
ファイルの先頭10行に対して、各行の1列目だけを抜き出して表示せよ。確認にはcutコマンドなどを用いよ。

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    print(re.split('\t', lines[i])[0])
```
