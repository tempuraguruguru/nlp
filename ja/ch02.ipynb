## 11. 先頭からN行を出力
ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

## 13. タブをスペースに置換
ファイルの先頭10行に対して、タブ1文字につきスペース1文字に置換して出力せよ。確認にはsedコマンド、trコマンド、もしくはexpandコマンドなどを用いよ。

```python
import re
```

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    line = lines[i]
    line = re.sub('\t', ' ', line)
    print(line)
```

## 14. 1列目を出力
ファイルの先頭10行に対して、各行の1列目だけを抜き出して表示せよ。確認にはcutコマンドなどを用いよ。

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for i in range(10):
    print(re.split('\t', lines[i])[0])
```

## 15. ファイルをN分割する
ファイルを行単位でN分割し、別のファイルに格納せよ。例えば、N=10としてファイルを10分割せよ。同様の処理をsplitコマンドで実現せよ。

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  count = 0
  lines = file.readlines()
  for i in range(0, len(lines), len(lines)//10):
    with open(f'/content/drive/MyDrive/nlp100/lesson02/data/split-{count}.txt', 'w') as file:
      file.write(''.join(lines[i:i+len(lines)//10]))
    count += 1
print(count)
```

## 16. ランダムに各行を並び替える
ファイルを行単位でランダムに並び替えよ（注意: 各行の内容は変更せずに並び替えよ）。同様の処理をshufコマンドで実現せよ。

```python
import random
```

```python
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  random.shuffle(lines)
  with open('/content/drive/MyDrive/nlp100/lesson02/data/random.txt', 'w') as file:
    file.write(''.join(lines))
```

## 17. １列目の文字列の異なり
1列目の文字列の異なり（文字列の種類）を求めよ。確認にはcut, sort, uniqコマンドを用いよ。

```python
import pandas as pd
```

## 18. 各行の1列目の文字列の出現頻度を求め、出現頻度の高い順に並べる
1列目の文字列の出現頻度を求め、出現頻度と名前を出現頻度の多い順に並べて表示せよ。確認にはcut, uniq, sortコマンドを用いよ。

```python
data = {'Name': [], 'Sex': [], 'Population': [], 'Year': []}
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for line in lines:
    line = re.sub('\n', '', line)
    line = re.split('\t', line)
    data['Name'].append(line[0])
    data['Sex'].append(line[1])
    data['Population'].append(line[2])
    data['Year'].append(line[3])

df = pd.DataFrame(data)
df = df.groupby('Name').count()
df['Count'] = df['Sex']
df = df[['Count']]
df = df.sort_values(by = 'Count', ascending = False)
df
```

## 19. 3列目の数値の降順に各行を並び替える
3列目の数値の逆順でファイルの各行を整列せよ（注意: 各行の内容は変更せずに並び替えよ）。同様の処理をsortコマンドで実現せよ。

```python
data = {'Name': [], 'Sex': [], 'Population': [], 'Year': []}
with open('/content/drive/MyDrive/nlp100/lesson02/popular-names.txt', 'r') as file:
  lines = file.readlines()
  for line in lines:
    line = re.sub('\n', '', line)
    line = re.split('\t', line)
    data['Name'].append(line[0])
    data['Sex'].append(line[1])
    data['Population'].append(int(line[2]))
    data['Year'].append(int(line[3]))

df = pd.DataFrame(data)
df = df.sort_values(by = 'Population', ascending = False)
df
```
